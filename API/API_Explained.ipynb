# Chapter 1 — The Restaurant Where Everything Began 🍛

Imagine you walk into a restaurant. You're hungry. You sit down at a table.

Where is the food made? In the **kitchen**. Can you walk into the kitchen, open the fridge, and cook your own biryani? No. The kitchen has its own rules, its own secret recipes, its own chefs.

So what do you do? You call the **waiter**.

- You tell the waiter: *"One paneer butter masala, please."*
- The waiter walks to the kitchen and tells the chef.
- The chef cooks.
- The waiter brings the food back to you.

You never saw the kitchen. You didn't need to. You only needed to know **how to talk to the waiter** — the menu tells you what you can order.

### Now replace the characters

| In the restaurant | In the software world |
|---|---|
| You (the customer) | **Client** — your app, website, or Python script |
| The waiter | **API** |
| The kitchen | **Server** — where data and logic live |
| The menu | **API documentation** — what you can ask for |
| Your order | **Request** |
| The food you get back | **Response** |

## So, what is an API?

**API = Application Programming Interface.**

In plain words: *an API is a set of rules that lets one piece of software talk to another piece of software.*

- **Application** → any software
- **Programming** → used by programmers/code
- **Interface** → the meeting point between two things

When you open a weather app on your phone, the app doesn't own a satellite. It politely asks a weather server via an API: *"What's the temperature in Bengaluru?"* — and the server replies with the data.

Let's actually watch that happen. Here we call a real, free API that returns a random joke... well, a random "to-do item" from a practice server:

````python
import requests

# We (client) ask the server (kitchen) through the API (waiter)
response = requests.get("https://jsonplaceholder.typicode.com/todos/1")

print("Status:", response.status_code)   # 200 means "order delivered successfully"
print("Data  :", response.json())        # the 'food' we got back
````

That little output is the whole idea of an API: **you asked, the server answered.** Everything else in this notebook is just details about *how* to ask properly and *how* to build a waiter of your own.

# Chapter 2 — Why Do We Even Need Them? 🤔

Let's imagine a world **without** APIs.

You build a travel app. You want to show flights, hotel prices, weather, maps, and let people pay. Without APIs you'd have to:

- build your own airline database 😵
- launch your own satellites for maps 🛰️
- become a bank to process payments 🏦

Impossible. Instead, you **borrow** these abilities from people who already built them.

### The real reasons APIs exist

1. **Reuse, don't reinvent.** Google already built maps. Razorpay/Stripe already built payments. You plug them in.
2. **Connect different systems.** Your Python backend, your Android app, and a React website can all talk to the *same* server through the same API.
3. **Hide complexity (and keep things safe).** The waiter doesn't give you the kitchen keys. An API exposes only what you're allowed to touch — not the entire database.
4. **Automation.** Instead of a human clicking buttons, code can do it: post a tweet, send an SMS, fetch stock prices every minute.
5. **Teams can work independently.** The frontend team and backend team agree on an API "contract" and then work in parallel.

### You use APIs every single day (without noticing)

- 🌦️ Weather app → weather API
- 🚕 Ola/Uber showing the map → Maps API
- 🔐 "Login with Google" button → OAuth API
- 💳 Paying on Swiggy/Zomato → payment API
- 📺 YouTube video embedded in a blog → YouTube API

> **One-line summary:** APIs are the reason software can *cooperate* instead of living on lonely islands.

# Chapter 3 — The Many Faces of APIs 🎭

APIs come in different "flavours". There are two useful ways to look at them: **who can use them** and **how they talk**.

## A) By who can use them (access)

| Type | Meaning | Example |
|---|---|---|
| **Public / Open API** | Anyone can use it (often with a free key) | OpenWeather, NASA, PokéAPI |
| **Private / Internal API** | Used only inside a company | Your company's internal HR system |
| **Partner API** | Shared with selected business partners | Airline data shared with travel agencies |

## B) By *style* of communication (the technical flavours)

### 1. REST (the most popular — we'll focus here)
Uses normal web URLs and HTTP methods (GET, POST, PUT, DELETE). Mostly returns **JSON**. Simple, readable, everywhere.
*Story:* a menu where each dish has its own page number — you open the exact page you want.

### 2. SOAP
An older, very strict style that uses **XML** with heavy rules. Still used in banks and enterprise systems.
*Story:* a formal government form — lots of fields, everything must be perfect.

### 3. GraphQL
You send **one query** and describe *exactly* which fields you want. No extra data, no missing data.
*Story:* telling the waiter, "Just the paneer, no gravy, extra butter" instead of the whole thali.

### 4. gRPC
Very fast, uses compact binary data. Popular for microservices talking to each other.
*Story:* chefs in the kitchen shouting short code words to each other — fast but not for customers.

### 5. WebSocket (real-time)
A connection that **stays open** so both sides can send messages anytime. Chat apps, live scores, stock tickers.
*Story:* not ordering by waiter at all — a live phone call with the kitchen.

### Bonus: Webhooks
Instead of *you* asking again and again ("Is it ready? Is it ready?"), the server **calls you** when something happens (e.g., "Payment successful!").

---
> 🎯 For beginners: learn **REST** first. 90% of the APIs you'll meet are REST APIs, and everything else becomes easier after.

## REST's four favourite verbs (HTTP methods)

| Method | Meaning | Restaurant meaning | Typical use |
|---|---|---|---|
| **GET** | Read data | "Show me the menu" | Fetch a list/item |
| **POST** | Create new data | "Place a new order" | Sign up, add a book |
| **PUT / PATCH** | Update existing data | "Change my order" | Edit profile |
| **DELETE** | Remove data | "Cancel my order" | Delete a post |

(These four are often called **CRUD**: Create, Read, Update, Delete.)

# Chapter 4 — Anatomy of a Request 🔬

Time to open up one single "order" and see what it's made of. Every API request has a few key parts. Let's meet them one by one.

## 4.1 The URL — the address 📍

A URL tells the request **where to go**. Let's cut one into pieces:

```
https://api.example.com:443/v1/books/42?lang=en&limit=5
└─┬──┘   └──────┬──────┘└┬┘ └───┬────┘ └──────┬───────┘
scheme       host      port    path        query string
```

- **Scheme** → `https` (secure) or `http`
- **Host / domain** → which server (`api.example.com`)
- **Port** → the "door number" on that server (usually hidden: 443 for https, 80 for http)
- **Path** → *which resource* you want (`/v1/books/42` = "book number 42, version 1")
- **Query string** → extra filters after `?`, written as `key=value` joined by `&`

The base part (`https://api.example.com/v1`) is often called the **base URL**, and each path (`/books`, `/users`) is called an **endpoint**.

````python
from urllib.parse import urlparse, parse_qs

url = "https://api.example.com:443/v1/books/42?lang=en&limit=5"
parts = urlparse(url)

print("Scheme :", parts.scheme)
print("Host   :", parts.hostname)
print("Port   :", parts.port)
print("Path   :", parts.path)
print("Query  :", parse_qs(parts.query))
````

### Query parameters in action
Instead of writing `?userId=1` by hand, Python's `requests` lets you pass a dictionary with `params`. Here we ask for only the posts written by user 1:

````python
import requests

r = requests.get("https://jsonplaceholder.typicode.com/posts", params={"userId": 1})

print("Final URL called:", r.url)
print("Number of posts :", len(r.json()))
print("First title     :", r.json()[0]["title"])
````

## 4.2 The Method — the verb 🗣️
We met them already: GET, POST, PUT/PATCH, DELETE. The method tells the server *what you want to do* with the resource at that URL.

## 4.3 Headers — the sticky notes on the envelope ✉️

Think of your request as a letter. The **body** is the letter inside. **Headers** are notes written on the outside of the envelope: who you are, what language you speak, what format you're sending.

Headers are **key: value** pairs. Common ones:

| Header | What it says |
|---|---|
| `Content-Type` | "The body I'm sending is in this format" (e.g., `application/json`) |
| `Accept` | "Please reply to me in this format" |
| `Authorization` | "Here's my proof of identity" (next chapter!) |
| `User-Agent` | "This is who/what is making the request" |

Let's use a handy testing site, **httpbin.org**, that simply *echoes back* whatever we send — perfect for seeing our own headers.

````python
headers = {
    "Accept": "application/json",
    "X-My-Custom-Header": "hello-from-my-notebook"
}

r = requests.get("https://httpbin.org/headers", headers=headers)
print(r.json())
````

See `X-My-Custom-Header` in the reply? The server received our sticky note. ✅

## 4.4 The Request Body — the letter itself 📨

When you **GET** something, you usually send no body: you're just asking.

But when you **POST** or **PUT**, you're handing over *data* — a new user, a new blog post, a new order. That data travels in the **request body**, most commonly as **JSON**.

**JSON** looks just like a Python dictionary:

```json
{
  "title": "My first post",
  "body": "Learning APIs is fun",
  "userId": 1
}
```

````python
new_post = {
    "title": "My first post",
    "body": "Learning APIs is fun",
    "userId": 1
}

# json=... automatically converts the dict to JSON
# and sets the Content-Type header for us
r = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_post)

print("Status:", r.status_code)   # 201 = Created
print("Server replied with:", r.json())
````

Notice the server replied with our data **plus an `id`** — it "saved" our post and gave it a number. (This practice server doesn't *really* store it, but it behaves as if it did.)

## 4.5 The Response — what comes back 📦

A response has the same style of parts as a request:

1. **Status code** — a number saying how it went
2. **Headers** — notes from the server
3. **Body** — the actual data (usually JSON)

### Status codes: the server's mood ring 🚦

| Range | Meaning | Examples |
|---|---|---|
| **1xx** | Info | (rarely seen) |
| **2xx** | ✅ Success | `200 OK`, `201 Created`, `204 No Content` |
| **3xx** | ↪️ Redirect | `301 Moved Permanently` |
| **4xx** | ❌ *You* made a mistake | `400 Bad Request`, `401 Unauthorized`, `403 Forbidden`, `404 Not Found`, `429 Too Many Requests` |
| **5xx** | 💥 *Server* has a problem | `500 Internal Server Error`, `503 Service Unavailable` |

> **Easy memory trick:** 2xx = "all good", 4xx = "your fault", 5xx = "our fault".

````python
r = requests.get("https://jsonplaceholder.typicode.com/posts/1")

print("Status code :", r.status_code)
print("Content-Type:", r.headers["Content-Type"])
print("Body (text) :", r.text[:100], "...")
print("Body (JSON) :", r.json()["title"])

# A page that doesn't exist:
bad = requests.get("https://jsonplaceholder.typicode.com/posts/99999")
print("\nAsking for a missing post gives:", bad.status_code)   # 404
````

# Chapter 5 — The Bouncer at the Door 🕴️ (Authorization)

Back to our restaurant. Some places are open to everyone. But the **VIP lounge** has a bouncer who asks: *"Who are you? Are you allowed in?"*

APIs have bouncers too. Two words you'll hear a lot — and they are *different*:

- **Authentication** → *"Who are you?"* (proving your identity)
- **Authorization** → *"What are you allowed to do?"* (permissions)

Example: A hotel lets you in with your room key (authentication), but that key opens *only your room*, not every room (authorization).

In practice, people often say "authorization" loosely for the whole thing, and it's usually sent in the **`Authorization` header**. Here are the common ways:

### 1. API Key 🔑
A long secret string given to you when you sign up. You send it in a header (or sometimes in the URL).
```
X-API-Key: abc123xyz
```
Simple and common (weather APIs, maps APIs).

### 2. Basic Auth 👤
You send `username:password`, encoded in Base64, in the header. Easy, but only safe over HTTPS.
```
Authorization: Basic dXNlcjpwYXNz
```

### 3. Bearer Token (e.g., JWT) 🎟️
You log in once, get a **token** (like a concert wristband), and show it with every request.
```
Authorization: Bearer eyJhbGciOiJIUzI1NiIs...
```
"Bearer" literally means *whoever bears (holds) this token gets access* — so guard it!

### 4. OAuth 2.0 🤝
Used for "Login with Google/GitHub". You let an app access *some* of your data **without giving it your password**. The app receives a limited-time token instead. It's more complex, but the final request still uses a Bearer token.

> 🛡️ **Golden rules**
> - Never put secret keys directly in code you upload to GitHub. Use environment variables.
> - Always use **HTTPS**.
> - Treat tokens like passwords.

### Let's see it in code
httpbin has special endpoints that act like a bouncer. First, **without** a token:

````python
r = requests.get("https://httpbin.org/bearer")
print("Without token:", r.status_code)   # 401 - bouncer says no
````

````python
headers = {"Authorization": "Bearer my-super-secret-token"}
r = requests.get("https://httpbin.org/bearer", headers=headers)
print("With token   :", r.status_code)   # 200 - welcome in!
print(r.json())
````

````python
# Basic Auth example (username: "bob", password: "pass123")
r = requests.get("https://httpbin.org/basic-auth/bob/pass123", auth=("bob", "pass123"))
print(r.status_code, r.json())
````

# Chapter 6 — Meet the Sidekicks: curl & Postman 🦸

Before writing a full program, developers like to *poke* an API quickly to see how it behaves. Two famous tools help with that.

## 🔧 curl — the command-line Swiss army knife

**curl** ("Client URL") sends requests from your terminal. It comes pre-installed on most Mac/Linux machines and modern Windows.

**Pros:** fast, scriptable, works anywhere. **Cons:** commands get long and less friendly.

Common flags:

| Flag | Meaning |
|---|---|
| `-X` | HTTP method (`GET`, `POST`...) |
| `-H` | add a header |
| `-d` | send body data |
| `-i` | show response headers too |
| `-v` | verbose (show everything) |

**Examples (run these in your terminal, or in the cells below):**

```bash
# simple GET
curl https://jsonplaceholder.typicode.com/posts/1

# POST with headers and JSON body
curl -X POST https://jsonplaceholder.typicode.com/posts \
  -H "Content-Type: application/json" \
  -d '{"title":"hello","body":"from curl","userId":1}'

# with authorization
curl https://httpbin.org/bearer -H "Authorization: Bearer mytoken"
```

## 📮 Postman — the friendly visual toolbox

**Postman** is a desktop/web app where you test APIs by **clicking instead of typing**.

**What you can do with it:**
- Pick a method (GET/POST/...) from a dropdown and paste a URL
- Add **Params**, **Headers**, **Authorization**, and **Body** in separate tabs
- Press **Send** and see status code, time, headers, and pretty-printed JSON
- Save requests into **Collections** (folders) and share with your team
- Use **Environments** for variables like `{{base_url}}` and `{{token}}`
- Write simple test scripts to check responses automatically
- Auto-generate documentation and mock servers

**Your first Postman walk-through:**
1. Open Postman → click **New → HTTP Request**
2. Choose `GET`, paste `https://jsonplaceholder.typicode.com/posts/1`
3. Click **Send** — see the JSON below 🎉
4. Change to `POST`, go to **Body → raw → JSON**, paste a JSON body, and **Send**
5. Go to the **Authorization** tab to try API Key / Bearer Token

**Other tools in the same family:** Insomnia, Thunder Client (VS Code extension), Bruno, HTTPie, and Swagger UI / Redoc (for reading API docs interactively).

> 🧭 **When to use which?**
> - Quick one-liner or automation → **curl**
> - Exploring, debugging, and sharing with a team → **Postman**
> - Real application → write code (Python `requests`, JavaScript `fetch`, etc.)

# Chapter 7 — Talking to an API from Python 🐍

You've already used the `requests` library through this notebook. Let's gather the patterns into one tidy cheat sheet, and add **good manners**: handling errors and slow servers.

### The four verbs in code

````python
import requests

BASE = "https://jsonplaceholder.typicode.com"

# GET — read
print("GET   :", requests.get(f"{BASE}/posts/1").status_code)

# POST — create
print("POST  :", requests.post(f"{BASE}/posts", json={"title": "t", "body": "b", "userId": 1}).status_code)

# PUT — replace/update
print("PUT   :", requests.put(f"{BASE}/posts/1", json={"id": 1, "title": "new", "body": "new", "userId": 1}).status_code)

# PATCH — partial update
print("PATCH :", requests.patch(f"{BASE}/posts/1", json={"title": "only title changed"}).status_code)

# DELETE — remove
print("DELETE:", requests.delete(f"{BASE}/posts/1").status_code)
````

### Good manners: timeouts and error handling

Servers sometimes go slow or break. A good client never waits forever and never assumes success.

````python
import requests

try:
    r = requests.get("https://jsonplaceholder.typicode.com/posts/1", timeout=5)  # give up after 5 seconds
    r.raise_for_status()          # raises an error for 4xx / 5xx responses
    data = r.json()
    print("Got post titled:", data["title"])

except requests.exceptions.Timeout:
    print("The server took too long to answer.")
except requests.exceptions.HTTPError as e:
    print("Server said no:", e)
except requests.exceptions.RequestException as e:
    print("Something went wrong with the request:", e)
````

### Keeping secrets safe
Never hardcode real keys. Read them from environment variables:


# Chapter 8 — Building Our Own API (The Fun Part!) 🏗️

So far we were the *customer*. Now we become the **waiter and kitchen**.

We'll build a tiny **Library API** that manages books. It will support:

| Action | Method & URL |
|---|---|
| List all books | `GET /books` |
| Get one book | `GET /books/<id>` |
| Add a book (needs API key 🔑) | `POST /books` |
| Update a book | `PUT /books/<id>` |
| Delete a book | `DELETE /books/<id>` |

### Which Python tool?
- **Flask** — tiny, beginner-friendly, very readable. *(We'll use this.)*
- **FastAPI** — modern, fast, automatic documentation page, data validation. *(Shown after.)*
- **Django REST Framework** — big and powerful, for large projects.

### Step 1: Define the app and some starting data
Our "database" is just a Python list (real apps use PostgreSQL, MySQL, MongoDB, etc.).

````python
from flask import Flask, jsonify, request

app = Flask(__name__)

# Our pretend database
books = [
    {"id": 1, "title": "The Alchemist", "author": "Paulo Coelho"},
    {"id": 2, "title": "Wings of Fire", "author": "A.P.J. Abdul Kalam"},
]

API_KEY = "my-secret-key-123"   # in real life, load from an environment variable
````

### Step 2: The "read" endpoints (GET)
`@app.route(...)` is a **decorator** that says: *"when someone visits this URL, run this function."* Whatever the function returns becomes the response.

Notice `<int:book_id>` in the path — Flask captures that part of the URL and hands it to our function. Also notice we return **proper status codes** (404 when something isn't found).

````python
@app.route("/")
def home():
    return jsonify(message="Welcome to the Library API 📚")

@app.route("/books", methods=["GET"])
def list_books():
    return jsonify(books)

@app.route("/books/<int:book_id>", methods=["GET"])
def get_book(book_id):
    for book in books:
        if book["id"] == book_id:
            return jsonify(book)
    return jsonify(error="Book not found"), 404
````

### Step 3: Write endpoints (POST, PUT, DELETE)

For POST we also add our **bouncer**: the request must carry the correct `X-API-Key` header, else we answer `401 Unauthorized`. We also **validate** the body — if `title` is missing, we say `400 Bad Request`.

````python
def is_authorized():
    return request.headers.get("X-API-Key") == API_KEY

@app.route("/books", methods=["POST"])
def add_book():
    if not is_authorized():
        return jsonify(error="Unauthorized: missing or wrong API key"), 401

    data = request.get_json(silent=True)           # read the JSON body
    if not data or "title" not in data:
        return jsonify(error="'title' is required"), 400

    new_book = {
        "id": max([b["id"] for b in books], default=0) + 1,
        "title": data["title"],
        "author": data.get("author", "Unknown"),
    }
    books.append(new_book)
    return jsonify(new_book), 201                   # 201 = Created

@app.route("/books/<int:book_id>", methods=["PUT"])
def update_book(book_id):
    if not is_authorized():
        return jsonify(error="Unauthorized"), 401
    data = request.get_json(silent=True) or {}
    for book in books:
        if book["id"] == book_id:
            book["title"] = data.get("title", book["title"])
            book["author"] = data.get("author", book["author"])
            return jsonify(book)
    return jsonify(error="Book not found"), 404

@app.route("/books/<int:book_id>", methods=["DELETE"])
def delete_book(book_id):
    if not is_authorized():
        return jsonify(error="Unauthorized"), 401
    for book in books:
        if book["id"] == book_id:
            books.remove(book)
            return "", 204                          # 204 = Done, nothing to return
    return jsonify(error="Book not found"), 404
````

### Step 4: Switch the kitchen on 🔥

Normally you'd save the code in `app.py` and run `python app.py`, which blocks the terminal. Inside a notebook that would freeze everything, so we start the server in a **background thread**. It will listen at `http://127.0.0.1:5050`.

(`127.0.0.1` / `localhost` = "this very computer". Only you can reach this server.)

````python
import threading
from werkzeug.serving import make_server

server = make_server("127.0.0.1", 5050, app)
thread = threading.Thread(target=server.serve_forever, daemon=True)
thread.start()

print("✅ Library API is running at http://127.0.0.1:5050")
````

## Bonus: the same API in FastAPI ⚡

FastAPI is the modern favourite. Its superpowers: **type hints for automatic validation**, and a **free interactive documentation page** at `/docs`.

Save this as `main.py` in a normal folder (not meant to run inside this notebook):

```python
from fastapi import FastAPI, Header, HTTPException
from pydantic import BaseModel

app = FastAPI()
books = [{"id": 1, "title": "The Alchemist", "author": "Paulo Coelho"}]

class BookIn(BaseModel):
    title: str
    author: str = "Unknown"

@app.get("/books")
def list_books():
    return books

@app.get("/books/{book_id}")
def get_book(book_id: int):
    for b in books:
        if b["id"] == book_id:
            return b
    raise HTTPException(status_code=404, detail="Book not found")

@app.post("/books", status_code=201)
def add_book(book: BookIn, x_api_key: str = Header(None)):
    if x_api_key != "my-secret-key-123":
        raise HTTPException(status_code=401, detail="Unauthorized")
    new = {"id": len(books) + 1, **book.model_dump()}
    books.append(new)
    return new
```

# Chapter 10 — The Recap & What's Next 🌅

Let's retell the whole story in a few lines:

- An **API** is the waiter between a **client** and a **server**.
- We need APIs to **reuse** other people's work, **connect** systems, **hide complexity**, and **automate** things.
- APIs come in types: **REST** (most common), SOAP, GraphQL, gRPC, WebSocket — plus public/private/partner by access.
- A request has a **URL**, a **method**, **headers**, an optional **body** (and query params).
- A response has a **status code**, **headers**, and a **body** (mostly JSON).
- **Authorization** (API key, Basic, Bearer/JWT, OAuth) is the bouncer that checks who you are and what you can do.
- **curl** and **Postman** let you poke at APIs without writing a program.
- In **Python**, `requests` lets you *consume* APIs and **Flask / FastAPI** let you *build* them.

### One-page cheat sheet

| Concept | Remember it as |
|---|---|
| URL | The address of the dish |
| Method | What you want to do (GET/POST/PUT/DELETE) |
| Headers | Notes on the envelope |
| Body | The letter inside |
| Auth | The bouncer's ID check |
| Status code | The server's mood (2xx ✅, 4xx ❌ you, 5xx 💥 server) |
